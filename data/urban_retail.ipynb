## Data Cleaning and Transformation

```python
## Fill up missing values with appropriate parameters
urban_retail_analyse_df.fillna({
    'gender': 'Unknown',
    'age_group': 'Unknown',
    'customer_segment': 'Unknown',
    'payment_method': 'Unknown',
    'sales_channel': 'Unknown',
    'customer_rating': 0.0,
    'return_flag': 'Unknown'
}, inplace=True)
```

```python
## get the new information relating to the cleaned dataset
urban_retail_analyse_df.info()
```

```python
## Convert order_date from string to datetime object
urban_retail_analyse_df['order_date'] = pd.to_datetime(urban_retail_analyse_df['order_date'])

## Convert phonenumber to String

urban_retail_analyse_df['phone_number'] = urban_retail_analyse_df['phone_number'].astype(str)
```

```python
urban_retail_analyse_df.info()
```

## Tables

```python
# 1. Customer Dataframe - Drop duplicates so ech customer is unique

customer_df = urban_retail_analyse_df[['customer_id', 'customer_name', 'phone_number', 'gender', 'age_group', 'city', 'region', 'customer_segment'
]].copy().drop_duplicates().reset_index(drop=True)

# 2. Product Dataframe (Drop duplicate rows so each product variant is unique)
product_df = urban_retail_analyse_df[[
    'product_id', 'product_name', 'product_category', 'unit_price', 'discount_rate'
]].copy().drop_duplicates(subset=['product_id']).reset_index(drop=True)

# 3. Order Dataframe (Keeps granular items linked to the order and customer keys)
order_df = urban_retail_analyse_df[[
    'order_id', 'customer_id', 'product_id', 'order_date', 'order_status', 'quantity', 'order_revenue'
    ]].copy().reset_index(drop=True)

# 4. Payment_Channel Dataframe (Unique attributes mapped to the transactional token)
payment_channel_df = urban_retail_analyse_df[[
    'order_id', 'payment_method', 'sales_channel'
    ]].copy().drop_duplicates(subset=['order_id']).reset_index(drop=True)

# 5. Delivery Dataframe (Fulfillment tracking attributes per unique order)
delivery_df = urban_retail_analyse_df[[
    'order_id', 'delivery_status', 'delivery_days', 'delivery_fee'
    ]].copy().drop_duplicates(subset=['order_id']).reset_index(drop=True)

# 6. Experience Dataframe (Post-purchase evaluation insights tied to the order)
experience_df = urban_retail_analyse_df[[
    'order_id', 'customer_rating', 'return_flag'
    ]].copy().drop_duplicates(subset=['order_id']).reset_index(drop=True)
```

```python
# Save Tables to CSV (Paths corrected for notebook position)
customer_df.to_csv(r'cleaned_data/customers.csv', index=False)
product_df.to_csv(r'cleaned_data/products.csv', index=False)
order_df.to_csv(r'cleaned_data/orders.csv', index=False)
payment_channel_df.to_csv(r'cleaned_data/payment_channels.csv', index=False)
delivery_df.to_csv(r'cleaned_data/delivery.csv', index=False)
experience_df.to_csv(r'cleaned_data/experiences.csv', index=False)

```

## Data Loading

```python
## Connect to database

def get_db_connection():
    connection = psycopg2.connect(
        host = 'localhost',
        database = 'urban_retail',
        user = 'postgres',
        password = 'admin'
        
    )
    return connection
```

```python
conn = get_db_connection()
```

```python
def create_tables():
    conn = get_db_connection()
    cursor = conn.cursor()
    create_table_query = '''
                            CREATE SCHEMA IF NOT EXISTS urban_retail;
                            
                            DROP TABLE IF EXISTS urban_retail.customers CASCADE;
                            DROP TABLE IF EXISTS urban_retail.products CASCADE;
                            DROP TABLE IF EXISTS urban_retail.orders CASCADE;
                            DROP TABLE IF EXISTS urban_retail.payment_channels CASCADE;
                            DROP TABLE IF EXISTS urban_retail.delivery CASCADE;
                            DROP TABLE IF EXISTS urban_retail.experiences CASCADE;
                            
                            
                            
                            CREATE TABLE IF NOT EXISTS urban_retail.customers(
                                customer_id TEXT PRIMARY KEY,
                                customer_name TEXT,
                                phone_number INT,
                                gender TEXT,
                                age_group TEXT,
                                city TEXT,
                                region TEXT,
                                customer_segment TEXT
                                
                            );
                            
                            
                            CREATE TABLE IF NOT EXISTS urban_retail.product(
                                product_id TEXT PRIMARY KEY,
                                product_name TEXT,
                                product_category INT,
                                unit_price INT,
                                discount_rate FLOAT
                                
                            );
                            
                            
                            CREATE TABLE IF NOT EXISTS urban_retail.orders(
                                order_id TEXT PRIMARY KEY,
                                customer_id TEXT,
                                product_id TEXT,
                                order_date TIMESTAMP WITHOUT TIME ZONE,
                                order_status TEXT,
                                quantity INT,
                                order_revenue FLOAT,
                                FOREIGN KEY (customer_id) REFERENCES urban_retail.customers(customer_id),
                                FOREIGN KEY (product_id) REFERENCES urban_retail.product(product_id)
                                
                            );
                            
                            
                            CREATE TABLE IF NOT EXISTS urban_retail.payment_channels(
                                order_id TEXT PRIMARY KEY,
                                payment_method TEXT,
                                sales_channel TEXT
                                
                            );
                            
                            CREATE TABLE IF NOT EXISTS urban_retail.delivery(
                                deliver_id TEXT PRIMARY KEY,
                                order_id TEXT,
                                delivery_status TEXT,
                                delivery_days FLOAT,
                                delivery_fee INT,
                                FOREIGN KEY (order_id) REFERENCES urban_retail.orders (order_id)
                            );
                            
                              
                            CREATE TABLE IF NOT EXISTS urban_retail.experience(
                                experience_id TEXT PRIMARY KEY,
                                order_id TEXT,
                                customer_rating FLOAT,
                                return_flag FLOAT,
                                FOREIGN KEY (order_id) REFERENCES urban_retail.orders (order_id)
                            );
    
    '''
    cursor.execute(create_table_query)
    conn.commit()
    cursor.close()
    conn.close()
    
```

```python
create_tables()
```

```python
conn = get_db_connection()

try:
    with conn.cursor() as cursor:
        cursor.execute("""
            ALTER TABLE urban_retail.experience
            ALTER COLUMN return_flag TYPE TEXT
            USING return_flag::TEXT;
        """)

    conn.commit()

except Exception:
    conn.rollback()
    raise

finally:
    conn.close()
```

## Insert Data into Database

```python
import csv

def load_data_from_csv(csv_path):
    conn = get_db_connection()
    cursor = conn.cursor()
    with open(csv_path, 'r') as file:
        reader = csv.reader(file)
        next(reader)
        for row in reader:
            cursor.execute(
                '''
                    INSERT INTO urban_retail.experience (order_id, customer_rating, return_flag)
                    VALUES (%s, %s, %s)
                ''',
                row
            )
            
    conn.commit()
    cursor.close()
    conn.close()
    
csv_file_path = r'cleaned_data/experiences.csv'

load_data_from_csv(csv_file_path)


```

## Insert data into Delivery Table

```python


def load_data_from_csv(csv_path):
    conn = get_db_connection()

    try:
        with conn.cursor() as cursor:
            with open(
                csv_path, "r", encoding="utf-8-sig", newline=""
            ) as file:
                reader = csv.reader(file)
                next(reader, None)

                for row in reader:
                    # Convert empty CSV fields to SQL NULL.
                    row = [
                        None if value.strip() == "" else value
                        for value in row
                    ]

                    cursor.execute(
                        """
                        INSERT INTO urban_retail.delivery (
                            order_id,
                            delivery_status,
                            delivery_days,
                            delivery_fee
                        )
                        VALUES (%s, %s, %s, %s)
                        """,
                        row
                    )

        conn.commit()
        print("Delivery data loaded successfully.")

    except Exception:
        conn.rollback()
        raise

    finally:
        conn.close()


load_data_from_csv("cleaned_data/delivery.csv")
```
